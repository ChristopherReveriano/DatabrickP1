##Q1. Combine Two Tables

```python
from pyspark.sql import Row

# Create the person data
person_data = [
    Row(personId=1, lastName="Wang", firstName="Allen"),
    Row(personId=2, lastName="Alice", firstName="Bob")
]

# Create Person DataFrame
person = spark.createDataFrame(person_data)

# Create the address data
address_data = [
    Row(addressId=1, personId=2, city="New York City", state="New York"),
    Row(addressId=2, personId=3, city="Leetcode", state="California")
]

# Create Address DataFrame
address = spark.createDataFrame(address_data)

# Display both DataFrames
print("Person Table:")
person.show()
print("\nAddress Table:")
address.show()
```

```python
df = person.join(address, person.personId == address.personId)
display(df)
```

### Q2. Employees Earning More Than Their Managers


```python
from pyspark.sql import Row

# Create the employee data
employee_data = [
    Row(id=1, name="Joe", salary=70000, managerId=3),
    Row(id=2, name="Henry", salary=80000, managerId=4),
    Row(id=3, name="Sam", salary=60000, managerId=None),
    Row(id=4, name="Max", salary=90000, managerId=None)
]

# Create Employee DataFrame
employee = spark.createDataFrame(employee_data)

# Display the DataFrame
employee.show()
```

```python
e1 = employee.alias('e1')
e2 = employee.alias('e2')
df = e1.join(e2, e1.id == e2.managerId,"inner" )
df.select(e1.name).where(e1.salary < e2.salary).show()
```

## Q1. Customer Placing the Largest Number of Orders

```python
# Create the orders data
orders_data = [
    Row(order_number=1, customer_number=1),
    Row(order_number=2, customer_number=2),
    Row(order_number=3, customer_number=3),
    Row(order_number=4, customer_number=3)
]

# Create DataFrame
orders = spark.createDataFrame(orders_data)

# Display the DataFrame
orders.show() 
```

```python
from pyspark.sql import functions as F

temp_df = orders.groupBy( orders.customer_number).count( )

max_num = temp_df.agg( F.max("count") ).collect()[0][0]


temp_df.select( temp_df.customer_number ) .where( F.col('count') == max_num ).show()
```

### Q2. Classes With at Least 5 Students

```python
from pyspark.sql import Row

# Create the courses data
courses_data = [
    Row(student="A", class_name="Math"),
    Row(student="B", class_name="English"),
    Row(student="C", class_name="Math"),
    Row(student="D", class_name="Biology"),
    Row(student="E", class_name="Math"),
    Row(student="F", class_name="Computer"),
    Row(student="G", class_name="Math"),
    Row(student="H", class_name="Math"),
    Row(student="I", class_name="Math")
]

# Create Courses DataFrame
courses = spark.createDataFrame(courses_data)

# Display the DataFrame
courses.show()
```

```python
from pyspark.sql import functions as F
t1 = courses.groupBy( courses.class_name ).count()
t1.select( F.col('class_name') ) .where( F.col('count') >= 5 ).show()
```
